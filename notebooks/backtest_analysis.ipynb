# Backtest Analysis

Template for evaluating Phase 1 edge signals with Phase 2 metrics.

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from backtesting import BacktestConfig, Backtester
from scripts.backtest import demo_signals
```

```python
signals = demo_signals()
result = Backtester(BacktestConfig(save_results=False)).run(signals)
bets = result.bets_frame()
result.rubric, result.report.metrics
```

```python
bets.assign(cumulative_pnl=bets["pnl"].cumsum()).plot(
    x="as_of", y="cumulative_pnl", title="PnL Curve"
)
plt.show()
```

```python
calibration = pd.DataFrame([bucket.model_dump() for bucket in result.report.calibration])
sns.lineplot(data=calibration, x="model_prob_mean", y="outcome_rate", marker="o")
plt.plot([0, 1], [0, 1], linestyle="--", color="gray")
plt.title("Calibration Curve")
plt.show()
```
